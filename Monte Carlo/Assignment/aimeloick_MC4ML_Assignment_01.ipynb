#### b) **Binomial with parameters $(n, p)$**
---


 **Expectation**

The binomial distribution: $X \sim \mathrm{Binom}(n, p)$

$$
E[X] = \sum x P(X=x) = \sum_{x=0}^{n} x \binom{n}{x} p^x (1-p)^{n-x}
$$

We use the combinatorial identity:

$$
x \binom{n}{x} = n \binom{n-1}{x-1}
$$

Therefore:

$$
E[X] = n \sum_{x=1}^{n} \binom{n-1}{x-1} p \, p^{x-1} (1-p)^{n-x}
= n p \sum_{x=1}^{n} \binom{n-1}{x-1} p^{x-1} (1-p)^{n-1-(x-1)}
$$

However, the sum is equal to 1, therefore:

$$
\boxed{E[X] = n p}
$$


 **Variance**

We have:

$$
\mathrm{Var}(X) = E[X^2] - (E[X])^2
$$

Let us calculate $E[X^2]$:

$$
E[X^2] = \sum_{x=0}^{n} x^2 \binom{n}{x} p^x (1-p)^{n-x}
$$

We can rewrite $x^2$ as $x(x-1) + x$, which allows us to simplify:

$$
E[X^2] = \sum_{x=0}^{n} x(x-1) \binom{n}{x} p^x (1-p)^{n-x} + \sum_{x=0}^{n} x \binom{n}{x} p^x (1-p)^{n-x}
$$

The second term is $E[X] = np$.  

The first term can be simplified using the combinatorial identity:

$$
x(x-1) \binom{n}{x} = n(n-1) \binom{n-2}{x-2}
$$

Therefore:

$$
\sum_{x=0}^{n} x(x-1) \binom{n}{x} p^x (1-p)^{n-x} = n(n-1) p^2 \sum_{x=2}^{n} \binom{n-2}{x-2} p^{x-2} (1-p)^{n-x}
$$

However, the sum $$\sum_{x=2}^{n} \binom{n-2}{x-2} p^{x-2} (1-p)^{n-x}
=1$$ as sum of the probabilities of a binomial law $$\mathrm{Bin}(n-2, p)$$

 So the first term is $n(n-1)p^2$.  

Therefore:

$$
E[X^2] = n(n-1)p^2 + np
$$

And therefore:

$$
\mathrm{Var}(X) = E[X^2] - (E[X])^2 = n(n-1)p^2 + np - (np)^2 = np(1-p)
$$

$$
\boxed{\mathrm{Var}(X) = np(1-p)}
$$

 **Characteristic function**
$$
\phi_X(t) = E[e^{itX}] = \sum_{x=0}^{n} \binom{n}{x} p^x (1-p)^{n-x} e^{itx}
$$

We can rewrite the sum as:

$$
\phi_X(t) = \sum_{x=0}^{n} \binom{n}{x} (p e^{it})^x (1-p)^{n-x}
$$

Recall the classical binomial expansion:

$$
(a + b)^n = \sum_{x=0}^{n} \binom{n}{x} a^x b^{n-x}
$$
Then,

$$
\boxed{\phi_X(t) = (1 - p + p e^{it})^n}
$$

#### **(c) Gaussian (Normal).**
---


 **Expectation**
$$
X \sim \mathcal{N}(\mu, \sigma^2)
$$

$$
p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)
$$

$$
E[X] = \int_{-\infty}^{+\infty} x p(x) \, dx
= \frac{1}{\sqrt{2\pi\sigma^2}} \int_{-\infty}^{+\infty} x
\exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right) dx
$$

Let $y = x - \mu$, we have $dx = dy$ :
$$
E[X] = \int_{-\infty}^{+\infty}
\frac{y + μ}{\sqrt{2\pi\sigma^2}} e^{-\frac{y^2}{2\sigma^2}} dy
$$
then,
$$
E[X] = \int_{-\infty}^{+\infty}
\frac{y}{\sqrt{2\pi\sigma^2}} e^{-\frac{y^2}{2\sigma^2}} dy
+ \mu \int_{-\infty}^{+\infty}
\frac{1}{\sqrt{2\pi\sigma^2}} e^{-\frac{y^2}{2\sigma^2}} dy
$$

The first term is equal to 0 (because is a odd function over a symetric interval).  
Therefore :

$$
E[X] = \mu \int_{-\infty}^{+\infty} \frac{1}{\sqrt{2\pi\sigma^2}}
e^{-\frac{y^2}{2\sigma^2}} dy = \mu
$$


$$
\boxed{E[X] = \mu}
$$


 **Variance**

$$
\mathrm{Var}(X) = E[(X - \mu)^2]
$$

$$
\mathrm{Var}(X) = \int_{-\infty}^{+\infty}
\frac{(x-\mu)^2}{\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} \, dt
dt
$$
Let $t = \frac{x-\mu}{\sigma}$, $dx = \sigma dt$ :

$$
\mathrm{Var}(X) = \int_{-\infty}^{+\infty}
\frac{(\sigma t)^2}{\sqrt{2\pi}} e^{-\frac{t^2}{2}} \, dt
$$
$$
= \sigma^2 \int_{-\infty}^{+\infty} \frac{t^2}{\sqrt{2\pi}} e^{-\frac{t^2}{2}} dt
$$

knowing,

$$
\int_{-\infty}^{+\infty} \frac{t^2}{\sqrt{2\pi}} e^{-\frac{t^2}{2}} dt = \frac{1}{\sqrt{2\pi}}\int_{-\infty}^{+\infty} t^2 e^{-\frac{t^2}{2}} dt = \frac{1}{\sqrt{2\pi}}(\sqrt{2\pi})=  1
$$

Thus :

$$
\boxed{\mathrm{Var}(X) = \sigma^2}
$$


**Characteristic Function**

$$
\phi_X(t) = E[e^{itX}] = \int_{-\infty}^{+\infty}
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left(-\frac{(x-\mu)^2}{2\sigma^2} + itx \right) dx
$$

Let's factorize the exponenty:

$$
-\frac{(x-\mu)^2}{2\sigma^2} + itx
= \frac{-(x^2 - 2x\mu + \mu^2) + 2\sigma^2 itx}{2\sigma^2}
$$

$$
= \frac{-x^2 + 2x\mu - \mu^2 + 2\sigma^2 itx}{2\sigma^2}
$$
$$
= \frac{-x^2 + 2x(\mu + i\sigma^2 t) - \mu^2}{2\sigma^2}
$$
$$
= \frac{-x^2 + 2x\mu - \mu^2 + 2\sigma^2 itx}{2\sigma^2}
$$
$$
= \frac{-x^2 + 2x(\mu + i\sigma^2 t) -(\mu + i\sigma^2 t)^2 +(\mu + i\sigma^2 t)^2- \mu^2}{2\sigma^2}
$$

$$
= \frac{-[x - (\mu + i\sigma^2 t)]^2 + (\mu + i\sigma^2 t)^2 - \mu^2}{2\sigma^2}
$$

Thus:

$$
-\frac{(x-\mu)^2}{2\sigma^2} + itx
= -\frac{[x - (\mu + i\sigma^2 t)]^2}{2\sigma^2} + i\mu t - \frac{\sigma^2 t^2}{2}
$$

Substitute into the characteristic function integral:

$$
\phi_X(t)
= \int_{-\infty}^{+\infty}
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\Big(-\frac{[x - (\mu + i\sigma^2 t)]^2}{2\sigma^2} + i\mu t - \frac{\sigma^2 t^2}{2}\Big)\, dx
$$

Factor out constants:

$$
\phi_X(t)
= e^{i\mu t - \frac{1}{2}\sigma^2 t^2}
\int_{-\infty}^{+\infty}
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\Big(-\frac{[x - (\mu + i\sigma^2 t)]^2}{2\sigma^2}\Big)\, dx
$$

The integral is the standard Gaussian integral, which equals 1. Therefore, the characteristic function is:

$$
\boxed{\phi_X(t) = \exp\Big(i\mu t - \frac{1}{2}\sigma^2 t^2\Big)}
$$

#### **(d) Exponential**
---


**Expectation**
$$
p(x) = \lambda e^{-\lambda x}, \quad x \ge 0
$$
$$
E[X] = \int_{0}^{+\infty} x \lambda e^{-\lambda x} dx
$$

By integration by parts:
Let $u = x$, $dv = \lambda e^{-\lambda x} dx \Rightarrow v = - e^{-\lambda x}$,

$$
E[X] = \left[-x e^{-\lambda x}\right]_{0}^{\infty}
+ \int_{0}^{+\infty} e^{-\lambda x} dx
$$
Since  $$\left[-x e^{-\lambda x}\right]_{0}^{\infty} = 0 - 0 = 0$$
Knowing :
$$
\int_0^{\infty} e^{-\lambda x} dx = \left[ -\frac{1}{\lambda} e^{-\lambda x} \right]_0^{\infty} = 0 - \left(-\frac{1}{\lambda}\right) = \frac{1}{\lambda}
$$
Thus :
$$
E[X] = \left[-x e^{-\lambda x}\right]_0^{\infty} + \int_0^{\infty} e^{-\lambda x} \, dx
= 0 + \frac{1}{\lambda} = \frac{1}{\lambda}
$$
$$
\boxed{E[X]
= \frac{1}{\lambda}
}$$
 **Variance**
$$
\mathrm{Var}(X) = E[X^2] - (E[X])^2
$$

$$
E[X^2] = \int_{0}^{+\infty} x^2 \lambda e^{-\lambda x} dx
= \lambda \frac{2!}{\lambda^3} = \frac{2}{\lambda^2}
$$
Source: [Wikipedia — Gaussian integral](https://en.wikipedia.org/wiki/Gaussian_integral)

So,
$$
\mathrm{Var}(X) = \frac{2}{\lambda^2}  - \frac{1}{\lambda^2}
$$
$$
\boxed{\mathrm{Var}(X) = \frac{1}{\lambda^2}}
$$
**Characteristic Function**

$$
\phi_X(t) = E[e^{itX}] = \int_{0}^{+\infty} \lambda e^{-\lambda x} e^{itx} \, dx
$$

$$
\phi_X(t) = \int_{0}^{+\infty} \lambda e^{(-\lambda + it)x} \, dx = \left[ \frac{\lambda e^{(-\lambda + it)x}}{-\lambda + it} \right]_0^{+\infty}
$$
$$
Since \quad \lambda > 0
$$

$$
\phi_X(t) = \left[ \frac{\lambda e^{(-\lambda + it)x}}{-\lambda + it} \right]_0^{+\infty}
= \frac{\lambda}{-\lambda + it} (0 - 1)
= \frac{\lambda}{\lambda - it}.
$$

$$
e^{(-\lambda + it)x} = e^{-\lambda x}e^{itx} \to 0 \text{ when } x \to +\infty \text{ since } -\lambda < 0.
$$

$$
\boxed{\phi_X(t) = \frac{\lambda}{\lambda - it}}.
$$

### **Q2. Gaussian sums.**
#### **Let's show that the sum of two IID Gaussian random variables with mean $μ$ and variance  $\sigma^2$ Gaussian-distributed.**

Let $X_1$ and $X_2$ be independent and identically distributed Gaussian random variables with mean $\mu$ and variance $\sigma^2$, that is, $X_1, X_2 \sim \mathcal{N}(\mu, \sigma^2)$.

Let's consider the sum $S_2 = X_1 + X_2$.

The characteristic function of a Gaussian random variable $X \sim \mathcal{N}(\mu, \sigma^2)$ is given by

$$
\phi_X(t) = E[e^{itX}] = \exp\left(it\mu - \frac{1}{2} \sigma^2 t^2 \right).
$$

Since $X_1$ and $X_2$ are independent, the characteristic function of $S_2$  is the product of the characteristic functions of each:

$$
\phi_{S_2}(t) = E[e^{it(X_1 + X_2)}] = E[e^{it X_1}] \cdot E[e^{it X_2}] = (\phi_X(t))^2.
$$

So the Gaussian characteristic function gives

$$
\phi_{S_2}(t) = \left( \exp\left(it\mu - \frac{1}{2} \sigma^2 t^2\right) \right)^2 = \exp\left(2 i \mu t - \sigma^2 t^2\right).
$$
This corresponds to the characteristic function of a Gaussian distribution with mean $2\mu$ and variance $2\sigma^2$. By the uniqueness property of characteristic functions, the distribution of $S_2$ is entirely determined by this function, therefore $S_2 \sim \mathcal{N}(2\mu, 2\sigma^2)$.

#### **Let's generalise to a sum of n IID Gaussian random variables.**
For a number $n$ of independent identically distributed Gaussian random variables $X_1, \dots, X_n \sim \mathcal{N}(\mu, \sigma^2)$. The characteristic function of the sum $S_n = X_1 + \dots + X_n$ is

$$
\phi_{S_n}(t) = \prod_{i=1}^n \phi_{X_i}(t) = (\phi_X(t))^n = \exp\left( i n \mu t - \frac{1}{2} n \sigma^2 t^2 \right),
$$

which is again the characteristic function of:

$$
S_n \sim \mathcal{N}(n \mu, n \sigma^2).
$$

Thus, the sum of independent Gaussian random variables is always Gaussian, and the mean and variance simply add.

### Q4. **Histogram function.**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def myhistogram(v, a, b, n):
    L = b - a
    bins_length = L / n
    bins = [a + i * bins_length for i in range(n + 1)]
    counts = np.zeros(n, dtype=int)

    for x in v:
        if a <= x < b:
            bin_index = int((x - a) / L * n)
            counts[bin_index] += 1
        elif x == b:
            counts[-1] += 1

    return counts, bins
```

```python
# Let's create a data like in the demo
data=np.array([np.random.rand() for i in range(1000)])

a, b = np.min(data), np.max(data)
# Let put n =100
n = 20

# Our function myhistogram
my_counts, my_bins = myhistogram(data, a, b, n)

# Function histogramn of python
python_counts, python_bins = np.histogram(data, bins=n)

print("My histogram counts:   ", my_counts)
print("Python histogram counts:", python_counts)
print("Same result:", np.allclose(my_counts, python_counts))

# Visualisation
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.bar(my_bins[:-1], my_counts, width=my_bins[1]-my_bins[0], align='edge', color='skyblue', edgecolor='black')
plt.title("My Histogram")
plt.xlabel("Value")
plt.ylabel("Count")

plt.subplot(1,2,2)
plt.bar(python_bins[:-1], python_counts, width=python_bins[1]-python_bins[0], align='edge', color='g', edgecolor='black')
plt.title("NumPy Histogram")
plt.xlabel("Value")

plt.tight_layout()
plt.show()
```

### Q5. **Non-uniform variates.**

#### (a) **Random choice in a list** $[1, 2, \dots, n]$ **with probabilities** $p_1, p_2, \dots, p_n$

Let use the Transformation (Inverse CDF) Method:

The cumulative distribution function (CDF) is:

$$
F(k) = P(X \le k) = \sum_{i=1}^{k} p_i
$$

Firstly we will generate a uniform random number:$$U \sim \text{Uniform}(0,1)$$  
Thus, applying the inverse cumulative distribution function involves choosing the smallest \(k\) such that $$F(k) \ge U$$. Then $$X = k$$.

---


```python
# Let's put n = 5
probs = np.array([0.05, 0.15, 0.30, 0.10, 0.40])  # exemple of probabilities for values 1 to 5
positions = np.arange(1, len(probs) + 1) #values 1 to 5
# Number of samples
n_samples = 10000
probs /= probs.sum()
cdf = np.cumsum(probs)
u = np.random.rand(n_samples)
# Inverse function cdf
def inverse_cdf(cdf, u):
  for k in range(len(cdf)):
    if u <= cdf[k]:
      return k+1

samples = np.array([inverse_cdf(cdf, ui) for ui in u])  # to get the inverse of the cdf of each u

# The empirical frequencies
values, counts = np.unique(samples, return_counts=True)
empirical = counts / n_samples

# Let's plot the  histogram (empirical vs theoretical)


plt.figure(figsize=(8, 5))
plt.bar(positions, empirical, width=0.6, alpha=0.8, label="Sampled")
plt.plot(positions, probs, 'r-', marker='o', lw=2, label="Theoretical PDF $p_i$")
plt.xticks(positions)
plt.xlabel("Value")
plt.ylabel("Probability")
plt.title("Discrete Non-uniform Sampling")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

####(b) **Uniform over [-1, 1]**



We want to generate a random variable $X$ such that:

$X \sim U[-1, 1]$.

That means the probability density function (PDF) is:

$f(x) =
\begin{cases}
\frac{1}{2}, & -1 \le x \le 1, \\
0, & \text{otherwise}.
\end{cases}$

The cumulative distribution function (CDF) is:

$F_X(x) = \frac{x+1}{2}, \quad -1 \le x \le 1$


So the inverse of the cdf is
$F^{-1}(y) = 2y - 1, \quad y \in [0,1]$




```python
# Number of samples
n_samples = 100000
U = np.random.rand(n_samples)
X = 2 * U - 1 #
# Theoretical PDF (is constant = 1/2 between -1 and 1)
x_theo = np.linspace(-1, 1, 300)
f_theo = 0.5 * np.ones_like(x_theo)

# Plot of histogram vs theoretical PDF
plt.figure(figsize=(7,5))
plt.hist(X, bins=100, density=True, alpha=0.7, label="Sampled")
plt.plot(x_theo, f_theo, 'r-', lw=2, label="Theoretical PDF (1/2)")
plt.title("Uniform distribution U[-1,1]")
plt.xlabel("x")
plt.ylabel("Density")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

#### (c) **Exponential with parameter λ.**

As shown in class, the inverse of the CDF of the exponential distribution is like:

$$
X = F^{-1}(U) = -\frac{1}{\lambda} \ln(1 - U)
$$



```python
lambda_param = 2.0
# Number of samples
n_samples = 10000

# Generate uniform random numbers
U = np.random.rand(n_samples)

# Transform to exponential
X = -np.log(U) / lambda_param

# Plot histogram and theoretical PDF
plt.hist(X, bins=100, density=True, alpha=0.6, color='skyblue', label='Sampled')
x_vals = np.linspace(0, np.max(X), 500)
plt.plot(x_vals, lambda_param * np.exp(-lambda_param * x_vals), 'r', lw=2, label='Theoretical PDF')
plt.title("Exponential with parameter λ.")
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.show()
```

### **Q6. Box–Muller method.**

**As shown is class:**

$$

$$

$$
\Rightarrow R \text{ has Rayleigh distribution } \quad p(r) = r e^{-r^2/2}

$$
\Rightarrow \theta \sim U(0, 2\pi)

$$

Let's generate $$U_1 \sim U(0,1) \quad \text{and} \quad U_2 \sim U(0,1):$$
With :
$$
 \quad R = \sqrt{-2 \ln U_1}
$$
And


$$
  \quad \theta = 2\pi U_2
$$
We will have $$
X \sim N(0,1), \quad Y \sim N(0,1)

```python
from scipy.stats import norm

# Number of samples
n_samples = 10000

# Let s generate uniform random numbers
U1 = np.random.rand(n_samples)
U2 = np.random.rand(n_samples)

# X and Y from Box–Muller transform
X = np.sqrt(-2 * np.log(U1)) * np.cos(2 * np.pi * U2)
Y = np.sqrt(-2 * np.log(U1)) * np.sin(2 * np.pi * U2)

# Theoretical Gaussian
x_axis = np.linspace(-5, 5, 1000)
pdf = norm.pdf(x_axis, 0, 1)

# Let's plot our X
plt.figure(figsize=(8,4))
plt.hist(X, bins=100, density=True, alpha=0.6, color='skyblue', label='X (Box–Muller)')
plt.plot(x_axis, pdf, 'r-', lw=2, label='N(0,1) PDF')
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Box–Muller Gaussian Sampling: X')
plt.legend()
plt.show()

# Let's plot our Y
plt.figure(figsize=(8,4))
plt.hist(Y, bins=100, density=True, alpha=0.6, color='lightgreen', label='Y (Box–Muller)')
plt.plot(x_axis, pdf, 'r-', lw=2, label='N(0,1) ')
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Box–Muller Gaussian Sampling: Y')
plt.legend()
plt.show()
```

###  **Q7. Monte Carlo estimation of π.**



```python
# Number of samples
n_samples = 1000000
x = np.random.uniform(-1, 1, n_samples)
y = np.random.uniform(-1, 1, n_samples)
# Counts point over circle
n_points_circle = np.sum((x**2 + y**2) <= 1)

# estimate the value of pi
pi_est = 4 * n_points_circle / n_samples
print('Estimator value = ', pi_est)
print('Theoretical value = ', np.pi)
```

```python
mean_est_err = 4 * np.std(n_points_circle) / np.sqrt(n_samples)  # std of Bernoulli / sqrt(L)
print('Estimator value = ', pi_est, "±", mean_est_err)
print('Absolute error = ', np.abs(pi_est - np.pi))

Lmax = 1000  # Number of cumulative points
pi_est_list = np.zeros(Lmax)
pi_est_err_list = np.zeros(Lmax)

s = 0  # Sum of indicator function
v = 0  # Sum of squared indicator function

for i in range(Lmax):
    x = np.random.uniform(-1, 1)
    y = np.random.uniform(-1, 1)
    r2 = x**2 + y**2
    indicator = 1.0 if r2 <= 1 else 0.0

    s += indicator
    v += indicator**2

    mean_est = 4 * s / (i+1)        # Cumulative estimator
    sec_mom_est = 4**2 * v / (i+1)  # Second moment scaled
    mean_est_err = np.sqrt(sec_mom_est - mean_est**2) / np.sqrt(i+1)  # The standard eror

    pi_est_list[i] = mean_est
    pi_est_err_list[i] = mean_est_err

# Plot with error bars
plt.figure(figsize=(10,5))
plt.errorbar(range(Lmax), pi_est_list, pi_est_err_list, errorevery=10, alpha=0.7, label='Estimate ± error')
plt.plot(range(Lmax), np.full(Lmax, np.pi), 'r--', label='π')
plt.xlabel('Number of points')
plt.ylabel('Estimated π')
plt.title('Monte Carlo Estimation of π: Convergence with Error Bars')
plt.legend()
plt.show()

# Like in the demo statisticalestimation
plt.figure(figsize=(10,5))
plt.plot(range(Lmax), pi_est_list, 'k-', label='Estimate')
plt.fill_between(range(Lmax), pi_est_list - pi_est_err_list, pi_est_list + pi_est_err_list,
                 alpha=0.5, edgecolor='#1B2ACC', facecolor='#089FFF', label='±1 std error')
plt.plot(range(Lmax), np.full(Lmax, np.pi), 'r--', label='π')
plt.xlabel('Number of points')
plt.ylabel('Estimated π')
plt.title('Monte Carlo Estimation of π: Convergence with Error Band')
plt.legend()
plt.show()
```

### Q8. **Two-state Markov chain.**

```python
# Parameters
alpha = 0.3   # P(0 to 1)
beta = 0.1    # P(1 to 0)
n_steps = 1000000

# Simulate trajectory
states = np.zeros(n_steps, dtype=int)  # start in state 0

for t in range(1, n_steps):
    if states[t-1] == 0:
        states[t] = 1 if np.random.rand() < alpha else 0
    else:
        states[t] = 0 if np.random.rand() < beta else 1

# Compute empirical occupation frequencies
p0_emp = np.mean(states == 0)
p1_emp = np.mean(states == 1)

# Theoretical stationary distribution
p0_theo = beta / (alpha + beta)
p1_theo = alpha / (alpha + beta)

print("Empirical P(0):", p0_emp)
print("Empirical P(1):", p1_emp)
print("Theoretical P(0):", p0_theo)
print("Theoretical P(1):", p1_theo)

# Visualize the trajectory (a short segment)
plt.figure(figsize=(8,3))
plt.plot(states[:200], drawstyle='steps-post')
plt.yticks([0,1])
plt.xlabel('Time step')
plt.ylabel('State')
plt.title('Two-State Markov Chain Trajectory (first 200 steps)')
plt.show()


```
